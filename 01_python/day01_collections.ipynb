## 2. tuple(튜플)

튜플은 여러 데이터를 순서대로 저장하는 자료구조.

- 리스트와 달리 생성 후 요소를 수정할 수 없음.
- 소괄호 ()를 사용하여 생성.
- 인덱싱과 슬라이싱이 가능.
- 이미지의 너비와 높이 같은 값을 묶어 관리할 때 활용.


```python
# 이미지 크기(너비, 높이)
image_size = (640,480)

print(image_size)

#튜플의 요소 접근
print(image_size[0])
print(image_size[1])

#튜플 언패킹
width, height = image_size

print("너비: ", width)
print("높이: ", height)
```

```python
#리스트는 수정 가능
image_list = [640,480]
image_list[0] = 800

print(image_list)

#튜플은 수정 불가능
image_tuple = (640,480)
image_tuple[0] = 800
```

## 3. Dictionary(딕셔너리)

Dictionary는 key-Value 쌍으로 데이터를 저장하는 자료구조

- 중괄호 {}를 사용
- key를 이용하여 원하는 Value에 접근
- 데이터를 추가, 수정, 삭제 가능
- cv에서 이미지의 파일명, 크기, 라벨 등을 관리할 때 활용

```python
image_info = {
  "filename": "cat01.jpg",
  "width": 640,
  "height": 480,
  "label": "cat"
}

print(image_info)
print(image_info["filename"])
print(image_info["width"])
print(image_info["label"])
```

```python
# 기존 데이터 수정
image_info["width"] = 800

# 새로운 데이터 추가
image_info["format"] = "JPEG"

print(image_info)
```

```python
dataset = [
  {"filename": "cat01.jpg", "label": "cat"},
  {"filename": "dog01.jpg", "label": "dog"},
  {"filename": "cat02.jpg", "label": "cat"},
]

print(dataset[0])
print(dataset[0]["filename"])
print(dataset[1]["label"])
```

## 4.Set(집합)

set은 중복을 허용하지 않는 자료구조이다.

- 중괄호 {}를 사용하여 생성할 수 있다.
- 중복된 값을 자동으로 제거한다.
- 순서가 없으며 인덱싱할 수 없다.
- 이미지 데이터셋에서 고유한 클래스 목록을 추출하 때 유용하다.

```python
labels = ["cat", "dog", "cat", "bird", "dog"]

unique_labels = set(labels)

print(labels)
print(unique_labels)
print(len(unique_labels))
```

```python
## 5. Practice - 이미지 데이터셋 분석
```

```python
dataset = [
  {"filename": "cat01.jpg", "size":(640,480),"label":"cat"},
  {"filename": "dog01.jpg", "size":(800,600),"label":"dog"},
  {"filename": "cat02.jpg", "size":(640,480),"label":"cat"},
  {"filename": "bird01.jpg", "size":(1280,720),"label":"bird"},
]

print(len(dataset))
print(dataset[0]["filename"])

labels =[]
for image in dataset:
  labels.append(image["label"])
unique_labels = set(labels)
print(unique_labels)

cat_count = 0
for image in dataset:
  if image["label"] =="cat":
    cat_count+=1
print("고양이 이미지 수:", cat_count)
```
